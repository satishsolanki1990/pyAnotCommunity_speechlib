```python
import os
from speechlib import Pipeline, PyAnnoteDiarizer, FasterWhisperASR


pipeline = Pipeline(
    diarization_model=PyAnnoteDiarizer(
        access_token=os.environ["HF_TOKEN"],
        min_speakers=1,
        max_speakers=2,
        community_model=True,
    ),
    asr_model=FasterWhisperASR("turbo"),
    language=None,          # None = auto-detect
    log_folder="logs",
    output_format="json",   # "txt", "json", or "both"
)
```

```python
from pyannote.audio import Pipeline
```

```python
# download the pipeline from Huggingface
pipeline = Pipeline.from_pretrained(
    "pyannote/speaker-diarization-community-1", 
    use_auth_token=os.environ["HF_TOKEN"])

# run the pipeline locally on your computer
output = pipeline("./sample.wav")

# print the predicted speaker diarization 
for turn, speaker in output.speaker_diarization:
    print(f"{speaker} speaks between t={turn.start:.3f}s and t={turn.end:.3f}s")
```
